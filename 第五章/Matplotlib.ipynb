```python
from IPython.core.pylabtools import figsize
from matplotlib.lines import lineStyles
from matplotlib.axes import Axes
from pydeck.data_utils.viewport_helpers import bbox_to_zoom_level

#matplotlib 是一个功能强大的数据可视化开源库
"""
画布:Figure
标题 : title
X轴:Xaxis
X刻度:Xtick
网格线:Grid
"""
import random
import matplotlib.pyplot as plt

#展示中文
plt.rcParams['font.sans-serif'] = ['SimHei']


x = [1,2,3,4,5,6,7]
y1= [random.randint(1,11) for i in range(7)]
y2 = [random.randint(1,20) for i in range(7)]

plt.figure(figsize=(10,5))
plt.plot(x,y1,label = '北京')
plt.plot(x,y2,label = '西安')


plt.title("气温变化折线图",fontsize=15)

plt.xlabel('时间')
plt.ylabel('温度')
plt.xticks(x)




plt.show()
```

```python
from matplotlib.lines import lineStyles
from matplotlib.pyplot import figure, axes
from matplotlib.axes import Axes

#画布对象,axes子图数组
fig, axes = plt.subplots(nrows=1, ncols=2, figsize=(20,6),dpi =100)


countries = ['中国', '美国', '印度', '沙特阿拉伯', '俄罗斯', '伊拉克', '伊朗']
values = [35,23,18,21,58,78,51]
axes1 : Axes = axes[0]

axes1.bar(countries,values,color ='g',width=0.6)
axes1.set_title("世界石油储备",fontsize=15)
axes1.set_xlabel("国家")
axes1.set_ylabel("石油储备")
axes1.grid()

countries1 = ['中国', '美国', '印度', '沙特阿拉伯', '俄罗斯', '伊拉克', '伊朗']
values = [35,23,18,21,58,78,51]
axes2 : Axes = axes[1]
axes2.pie(values, labels=countries1,autopct='%1.1f%%')
axes2.legend(loc = 'lower center',ncol=4)

plt.savefig('data/01.png')




```
